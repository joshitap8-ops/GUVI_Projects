```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import confusion_matrix, adjusted_rand_score
import seaborn as sns

iris = load_iris()
X = iris.data
y = iris.target

print("Dataset shape:", X.shape)
print("Flower species:", iris.target_names)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=3,random_state=42,n_init=10)
clusters = kmeans.fit_predict(X_scaled)

print("Predicted clusters:")
print(clusters)
```

```python
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 5))
plt.scatter(X_2d[:, 0],X_2d[:, 1],c=clusters,cmap="viridis",s=50)
plt.title("Iris Flower Clusters Using K-Means")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(label="Predicted Cluster")
plt.show()
```

```python
comparison = pd.DataFrame({"Actual Species": iris.target_names[y],"Predicted Cluster": clusters})
print(comparison.head(15))

ari = adjusted_rand_score(y, clusters)
print("Adjusted Rand Index:", round(ari, 3))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(X_2d[:, 0], X_2d[:, 1],c=y, cmap="viridis", s=40)
axes[0].set_title("Actual Flower Species")

axes[1].scatter(X_2d[:, 0], X_2d[:, 1],c=clusters, cmap="viridis", s=40)
axes[1].set_title("Predicted K-Means Clusters")

plt.tight_layout()
plt.show()
```
